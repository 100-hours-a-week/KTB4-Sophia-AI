## 3번 퀘스트
다음 1차원 배열에 새로운 차원을 추가하여 2차원 배열로 변환하고, 최종 배열의 차원 수를 출력하세요.
- newaxis를 사용하여 배열의 차원을 추가하고, 새로운 차원에서의 배열 모양을 확인한 후 최종 차원 수를 출력하세요.


```
import numpy as np
array = np.array([7, 14, 21])
```



````python
# 1. newaxis로 배열의 차원 추가
# 2. 새로운 차원에서의 배열 모양 확인
# 3. 최종 차원 수 출력
import numpy as np

# 원본 배열
vector = np.array([7, 14, 21])

# 차원 추가
column_vector = vector[:, np.newaxis]

print("원본 배열:", vector)

print("차원을 추가한 배열:\n", column_vector)

print("최종 차원 수", column_vector.ndim)
````

# 연산 - 미니퀘스트

## 1번 퀘스트


## 2번 퀘스트
다음 NumPy 배열에서 브로드캐스팅을 활용하여 각 행에 [1, 2, 3]을 더하는 코드를 작성하세요.

## 3번 퀘스트
주어진 2차원 NumPy 배열에 대해, 축(axis)을 기준으로 최댓값을 구하고 최종 배열의 차원을 출력하는 코드를 작성하세ㅔ요.


```
import numpy as np
array = np.array([[3, 7, 2], [8, 4, 6]])
```



````python
import numpy as np
array = np.array([[3, 7, 2], [8, 4, 6]])
max_axis0 = np.max(array, axis=0)
print(max_axis0)
print(max_axis0.ndim)

max_axis1 = np.max(array, axis=1)
print(max_axis1)
print(max_axis1.ndim)
````

# 유니버설 함수 - 미니퀘스트

## 1번 퀘스트
배열 [1, 2, 3, 4]에 대해 NumPy의 유니버설 함수를 사용하여 모든 요소를 제곱한 결과를 출력하는 코드를 작성하세요.

````python
import numpy as np

values = np.array([1, 2, 3, 4])
print("np.square(values):", np.square(values))
````

## 2번 퀘스트
다음 두 배열의 요소별 합을 계산하고, 결과를 새로운 배열이 아닌 기존 배열에 저장하는 코드를 작성하세요.


```
array1 = np.array([10, 20, 30])
array2 = np.array([1, 2, 3])
```



````python
import numpy as np

array1 = np.array([10, 20, 30])
array2 = np.array([1, 2, 3])

np.add(array1, array2, out=array1)
print(array1)
````

## 3번 퀘스트
다음 배열에 대해 NumPy 유니버설 함수를 사용하여 요소별로 자연로그(log)를 계산하고, 특정 조건(log 값 > 1)을 만족하는 요소만 출력하는 코드를 작성하세요.


```
array = np.array([1, np.e, 10, 100])
```



````python
import numpy as np

array = np.array([1, np.e, 10, 100])

result = array[np.log(array) > 1]
print(result)
````
