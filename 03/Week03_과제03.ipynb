
````python
# 샘플 데이터 (비정형 텍스트)
text = "안녕하세요!!! 저는 AI 모델-입니다. 12345 데이터를   정리해 보겠습니다."

targets = ['!', '-', '1', '2', '3', '4', '5', '.']

cleaned_text = text
for char in targets:
    cleaned_text = cleaned_text.replace(char,'')

cleaned_text = " ".join(cleaned_text.split())

print(cleaned_text)
````

## 3번 퀘스트
주어진 텍스트 데이터를 문장 단위로 분리한 후, 각 문장의 단어 개수를 데이터프레임으로 변환하는 코드를 작성하세요.


```
# 샘플 텍스트 데이터
text = "자연어 처리는 재미있다. 파이썬과 pandas를 활용하면 편리하다. 데이터 분석은 흥미롭다."
```



````python
import pandas as pd

text = "자연어 처리는 재미있다. 파이썬과 pandas를 활용하면 편리하다. 데이터 분석은 흥미롭다."

sentences = [s.strip() for s in text.split('.') if s.strip()]

result_data = []
for sentence in sentences:
    count = len(sentence.split())
    result_data.append({'문장': sentence, '단어개수':count})

df = pd.DataFrame(result_data)

print(df)
````


````python
# 데이터 생성
np.random.seed(10)
x = np.random.randn(50) * 2
y = np.random.randn(50) * 2
categories = np.random.choice(['A', 'B', 'C'], size=50)

colors = {'A': 'red', 'B': 'blue', 'C': 'green'}  # 카테고리별 색상 지정

for cat in np.unique(categories):
    idx = categories == cat
    plt.scatter(x[idx], y[idx],
                color=colors[cat],
                label=f'Category {cat}',
                alpha=0.7,
                s=80)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot by Category")
plt.legend()

plt.show()
````

plt.subplots()을 사용하여 2x1 형태의 서브플롯을 만들고, 첫 번째 서브플롯에는 y=x^2, 두 번째 서브플롯에는 y=x^3을 그리는 코드를 작성하세요.

````python
# 데이터 생성
x = np.linspace(-5, 5, 100)
y1 = x ** 2  # x의 제곱
y2 = x ** 3  # x의 세제곱

fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(8, 6))

# 첫 번째 서브플롯
axes[0].plot(x, y1, color='b', label='x^2')
axes[0].set_title("Square Function")
axes[0].set_xlabel("X-axis")
axes[0].set_ylabel("Y-axis")
axes[0].legend()

# 두 번째 서브플롯
axes[1].plot(x, y2, color='r', label='x^3')
axes[1].set_title("Cubic Function")
axes[1].set_xlabel("X-axis")
axes[1].set_ylabel("Y-axis")
axes[1].legend()

plt.tight_layout()

plt.show()
````

````python
import matplotlib.gridspec as gridspec

# 데이터 생성
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.random.randn(100)
categories = ['A', 'B', 'C', 'D', 'E']
values = [3, 7, 5, 2, 8]

fig = plt.figure(figsize=(10, 10))
gs = gridspec.GridSpec(3, 2, figure=fig)

# 1. 선 그래프
ax1 = fig.add_subplot(gs[0, :])
ax1.plot(x, y1, color='blue', label='sin(x)')
ax1.set_title("Line Plot")
ax1.legend()

# 2. 산점도
ax2 = fig.add_subplot(gs[1, 0])
ax2.scatter(x, y2, color='red', alpha=0.5)
ax2.set_title("Scatter Plot")

# 3. 막대 그래프
ax3 = fig.add_subplot(gs[1, 1])
ax3.bar(categories, values, color='green')
ax3.set_title("Bar Chart")

# 4. 히스토그램
ax4 = fig.add_subplot(gs[2, :])
ax4.hist(x, bins=30, color='purple', alpha=0.7)
ax4.set_title("Histogram")

plt.tight_layout()
plt.show()
````

```
# 세 개의 과일 집합 정의
set_A = {"사과", "바나나", "체리", "망고"}
set_B = {"바나나", "망고", "포도", "수박"}

```



````python
# 세 개의 과일 집합 정의
set_A = {"사과", "바나나", "체리", "망고"}
set_B = {"바나나", "망고", "포도", "수박"}
set_C = {"망고", "수박", "딸기", "오렌지"}

only_A = set_A - set_B - set_C
only_B = set_B - set_A - set_C
only_C = set_C - set_A - set_B

intersection_AB = set_A.intersection(set_B)
intersection_BC = set_B.intersection(set_C)
intersection_AC = set_A.intersection(set_C)

intersection_all = set_A & set_B & set_C

print(only_A, only_B, only_C)
print(intersection_AB, intersection_BC, intersection_AC)
print(intersection_all)
````

````python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 시계열 데이터 생성
np.random.seed(42)
date_range = pd.date_range(start="2023-01-01", periods=100, freq="D")  # 100일간의 날짜 생성
values = np.cumsum(np.random.randn(100))  # 랜덤 값의 누적합

df = pd.DataFrame({"Date": date_range, "Value": values})

plt.figure(figsize=(10, 5))
sns.lineplot(x="Date", y="Value", data=df, color="blue", marker="o")
plt.xlabel("Date")
plt.ylabel("Value")
plt.title("Time Series Line Plot")
plt.xticks(rotation=45)
plt.show()
````

## 3번 퀘스트
1번 퀘스트에서 생성한 시계열 데이터에서 이상치를 탐지하고, 이상치만 강조하여 그래프에 표시하는 코드를 작성하세요.

이상치는 사분위수 범위(IQR)를 이용해 판단합니다.


```
# 시계열 데이터 생성
np.random.seed(42)
date_range = pd.date_range(start="2023-01-01", periods=100, freq="D")
values = np.cumsum(np.random.randn(100))
```



````python
Q1 = df["Value"].quantile(0.25)
Q3 = df["Value"].quantile(0.75)
IQR = Q3 - Q1
k = 0.5
lower_bound = Q1 - k * IQR
upper_bound = Q3 + k * IQR

df["Outlier"] = (df["Value"] < lower_bound) | (df["Value"] > upper_bound)

outliers = df[df["Outlier"]]

plt.figure(figsize=(10, 5))
sns.lineplot(x="Date", y="Value", data=df, label="Original Data", color="blue")
sns.scatterplot(x="Date", y="Value", data=outliers, color="red", label="Outliers", s=100)

plt.xlabel("Date")
plt.ylabel("Value")
plt.title("Time Series with Outlier Detection")
plt.legend()
plt.xticks(rotation=45)
plt.show()
````

````python
# 샘플 데이터 생성
np.random.seed(42)
sample_data = np.random.normal(loc=50, scale=5, size=30)  # 평균 50, 표준편차 5인 데이터 30개 생성

test_value = 52
t_stat, p_value = stats.ttest_1samp(sample_data, popmean=test_value)

print(f"t-통계량: {t_stat}")
print(f"p-값: {p_value}")

if p_value < 0.05:
    print("p- 값이 0.05보다 작으므로 평균이 52와 유의미한 차이가 있음")
else:
    print("p- 값이 0.05보다 크므로, 평균이 52와 유의미한 차이가 있다고 보기 힘듦")
````

## 2번 퀘스트

```

observed = np.array([50, 60, 90])

```



````python
# 관측된 데이터 (Observed)
observed = np.array([50, 60, 90])

# 기대값 (Expected)
expected = np.array([66, 66, 66]) * (observed.sum() / np.sum([66, 66, 66]))

chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)

print(chi2_stat)
print(f"p- 값: {p_value}")

if p_value < 0.05:
    print("해석: p- 값이 0.05보다 작으므로, 관측값과 기대값 사이에 유의미한 차이가 있음")
else:
    print("해석: p- 값이 0.05보다 크므로, 관측값과 기대값 사이에 유의미한 차이가 있다고 보기 힘듦")
````

````python
# 데이터 생성 (광고 A와 B의 클릭 여부)
observed_data = pd.DataFrame({
    "Ad_A": [120, 380],  # 광고 A 클릭(120명) vs 미클릭(380명)
    "Ad_B": [150, 350]   # 광고 B 클릭(150명) vs 미클릭(350명)
}, index=["Click", "No Click"])

chi2, p_value, dof, expected = stats.chi2_contingency(observed_data)

print(f"카이제곱: {chi2:.4f}")
print(f"p-값: {p_value:.4f}")

if p_value < 0.05:
    print("해석: p-값이 0.05보다 작으므로, 두 광고의 클릭률에는 유의미한 차이가 있음")
else:
    print("해석: p-값이 0.05보다 크므로, 두 광고의 클릭률 차이가 유의미하다고 보기 힘듦")

click_rates = observed_data.loc["Click"] / observed_data.sum()
df_plot = click_rates.reset_index()
df_plot.columns = ["Ad", "Click_Rate"]

plt.figure(figsize=(6, 4))
sns.barplot(x="Ad", y="Click_Rate", data=df_plot, palette="viridis")
plt.title("Click-Through Rate (CTR) Comparison")
plt.ylim(0, 0.4)
plt.ylabel("Click Rate")
plt.show()
````
