```python
import pandas as pd
import seaborn as sns

# seaborn 내장 데이터셋 로드
df = sns.load_dataset('tips')
print(df.shape)
df.head()
```

## 과제 2.1

```python
grouped = df.groupby('day').agg(
    방문횟수=('total_bill', 'count'),
    평균식사금액=('total_bill', 'mean'),
    평균팁=('tip', 'mean')
).round(2)

print(grouped)
```

```python
# 성별 + 흡연 여부 복합 groupby
grouped2 = df.groupby(['sex', 'smoker'])['tip'].mean().round(2).reset_index()
grouped2.columns = ['성별', '흡연여부', '평균팁']
print(grouped2)
```

## 과제 2.2

```python
# 단일 조건: 팁이 5달러 이상인 손님
high_tip = df[df['tip'] >= 5]
print(f"팁 5달러 이상: {len(high_tip)}명")
print(high_tip.head())
```

```python
# 복합 조건: 저녁 식사 AND 일행 3명 이상 AND 비흡연자
filtered = df[(df['time'] == 'Dinner') & (df['size'] >= 3) & (df['smoker'] == 'No')]
print(f"저녁/3명 이상/비흡연: {len(filtered)}명")
print(filtered)
```
